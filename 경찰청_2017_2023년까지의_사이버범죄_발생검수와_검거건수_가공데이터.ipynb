```python
import pandas as pd

file1 = '경찰청_연도별 사이버 범죄 통계 현황_20231231.csv'
file2 = '경찰청_사이버사기범죄 현황_20241231.csv'

#1.데이터 로드
df1 = pd.read_csv(file1, encoding = 'EUC-KR')
df2 = pd.read_csv(file2, encoding = 'EUC-KR')

#칼럼명 앞뒤의 공백 제거
df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

#2.2017~2023년 데이터 필터링
df1_filtered = df1[(df1['연도'] >=2017) & (df1['연도'] <=2023)].copy()
df2_filtered = df2[(df2['연도'] >=2017) & (df2['연도'] <=2023)].copy()

# [추가] 두 데이터프레임을 하나로 통합 (행 방향으로 합치기)
df_filtered = pd.concat([df1_filtered, df2_filtered], ignore_index=False)

#3.연애빙자/사이버투자 관련행 제외
df_filtered = df_filtered[~df_filtered['구분'].str.contains('연애|연예|투자', na = False)].copy()


#4.그룹별로 묶을 컬럼 리스트 정의
hacking_cols = ['해킹_계정도용', '해킹_단순침입', '해킹_자료유출', '해킹_자료훼손']
scam_cols = ['직거래', '쇼핑몰', '게임', '이메일 무역', '사이버사기_직거래', '사이버사기_쇼핑몰',
       '사이버사기_게임', '사이버사기_이메일 무역사기']
cyber_cols = ['사이버금용범죄_피싱', '사이버금용범죄_메신저이용사기', '사이버금용범죄_기타', '사이버 음란물_아동음란물', '사이버 음란물_불법촬영물유포', '사이버 도박_스포츠토토', '사이버 도박_경마 경륜 경정',
       '사이버 도박_카지노', '사이버 도박_기타', '사이버 명예훼손_모욕', '사이버 스토킹']
Others_cols = ['사이버사기_기타', '기타 불법 컨텐츠']

#5. 데이터프레임에 실제로 존재하는 컬럼만 필터링
hacking_present = [c for c in hacking_cols if c in df_filtered.columns]
scam_present = [c for c in scam_cols if c in df_filtered.columns]
cyber_present = [c for c in cyber_cols if c in df_filtered.columns]
Others_present = [c for c in Others_cols if c in df_filtered.columns]


#그룹별 합산
df_filtered['Hacking'] = df_filtered[hacking_present].fillna(0).sum(axis=1) if hacking_present else 0
df_filtered['Scam'] = df_filtered[scam_present].fillna(0).sum(axis=1) if scam_present else 0
df_filtered['Cyber'] = df_filtered[cyber_present].fillna(0).sum(axis=1) if cyber_present else 0
df_filtered['Others'] = df_filtered[Others_present].fillna(0).sum(axis=1) if Others_present else 0

group_df = df_filtered.groupby(['연도', '구분'])[['Hacking', 'Scam', 'Cyber', 'Others']].sum().reset_index()

#6.새로운 통합 영문 데이터프레인 생성
mixing_df = pd.DataFrame()
mixing_df['Year'] = group_df['연도']
mixing_df['Type'] = group_df['구분']
mixing_df['Hacking'] =group_df['Hacking']
mixing_df['Scam'] = group_df['Scam']
mixing_df['Cyber'] = group_df['Cyber']
mixing_df['Others'] = group_df['Others']


#통합 컬럼 생성
#mixing_df['Year_Type'] = mixing_df['Year'].astype(str) + '_' + mixing_df['Type'].astype(str)
#8.정렬 수행
#mixing_df = mixing_df.sort_values(by = ['Year', 'Type'], ascending = [False, True]).reset_index(drop = True)

#9.최종적으로 필요한 컬럼만 슬라이싱
final_cols = ['Year','Type', 'Hacking', 'Scam', 'Cyber', 'Others']
mixing_df = mixing_df[final_cols]

#10. CSV 파일로 저장
mixing_df.to_csv("경찰청_사이버범죄_대그룹_통합_추출.csv", index = False, encoding = 'EUC-KR')

#결과
display(mixing_df)
```

```python
import pandas as pd

# 1. 파일 이름 정의
file1 = '경찰청_연도별 사이버 범죄 통계 현황_20231231.csv'
file2 = '경찰청_사이버사기범죄 현황_20241231.csv'

# 2. 데이터 로드
df1 = pd.read_csv(file1, encoding='EUC-KR')
df2 = pd.read_csv(file2, encoding='EUC-KR')

# 3. 컬럼명 앞뒤 공백 제거
df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

# 4. 2017~2023년 데이터 필터링
df1_filtered = df1[(df1['연도'] >= 2017) & (df1['연도'] <= 2023)].copy()
df2_filtered = df2[(df2['연도'] >= 2017) & (df2['연도'] <= 2023)].copy()

# 5. 두 파일의 데이터 결합
df_filtered = pd.concat([df1_filtered, df2_filtered], ignore_index=True)

# 6. '구분' 컬럼 앞뒤 공백 제거 및 특정 행 제외
df_filtered['구분'] = df_filtered['구분'].astype(str).str.strip()
df_filtered = df_filtered[~df_filtered['구분'].str.contains('연애|연예|투자', na=False)].copy()

# 7. 년도와 구분 기준으로 중복 행 원천 합산
df_grouped = df_filtered.groupby(['연도', '구분']).sum().reset_index()


# 8. 대그룹 컬럼 리스트 정의 및 반영
hacking_cols = ['해킹_계정도용', '해킹_단순침입', '해킹_자료유출', '해킹_자료훼손']
scam_cols = ['직거래', '쇼핑몰', '게임', '이메일 무역', '사이버사기_직거래', '사이버사기_쇼핑몰', '사이버사기_게임', '사이버사기_이메일 무역사기']
cyber_cols = ['사이버금용범죄_피싱', '사이버금용범죄_메신저이용사기', '사이버금용범죄_기타', '사이버금융범죄_피싱', '사이버금융범죄_메신저이용사기', '사이버금융범죄_기타',
              '사이버 음란물_아동음란물', '사이버 음란물_불법촬영물유포', '사이버 도박_스포츠토토', '사이버 도박_경마 경륜 경정',
              '사이버 도박_카지노', '사이버 도박_기타', '사이버 명예훼손_모욕', '사이버 스토킹']
Others_cols = ['사이버사기_기타', '기타 불법 컨텐츠']

hacking_present = [c for c in hacking_cols if c in df_grouped.columns]
scam_present = [c for c in scam_cols if c in df_grouped.columns]
cyber_present = [c for c in cyber_cols if c in df_grouped.columns]
Others_present = [c for c in Others_cols if c in df_grouped.columns]

mixing_df = pd.DataFrame()
mixing_df['년도'] = df_grouped['연도']
mixing_df['구분'] = df_grouped['구분']

mixing_df['Hacking'] = df_grouped[hacking_present].fillna(0).sum(axis=1) if hacking_present else 0
mixing_df['Scam'] = df_grouped[scam_present].fillna(0).sum(axis=1) if scam_present else 0
mixing_df['Cyber'] = df_grouped[cyber_present].fillna(0).sum(axis=1) if cyber_present else 0
mixing_df['Others'] = df_grouped[Others_present].fillna(0).sum(axis=1) if Others_present else 0


# 9. 세로형 변환 (Melt)
final_melted = mixing_df.melt(
    id_vars=['구분', '년도'],
    value_vars=['Hacking', 'Scam', 'Cyber', 'Others'],
    var_name='범죄종류',
    value_name='값'
)

# 10. 정렬 기준 설정 (년도 내림차순, 구분 오름차순, 범죄종류 순서 제어)
# 이미지와 똑같이 Hacking -> Scam -> Cyber -> Others 순서로 고정하기 위해 카테고리 지정
final_melted['범죄종류'] = pd.Categorical(final_melted['범죄종류'], categories=['Hacking', 'Scam', 'Cyber', 'Others'], ordered=True)
final_melted = final_melted.sort_values(by=['년도', '구분', '범죄종류'], ascending=[False, True, True]).reset_index(drop=True)


# ⭐⭐⭐ [핵심 변경: 원하시는 엑셀 그룹화/병합 형태로 레이아웃 수정] ⭐⭐⭐
# 구분과 년도를 행 인덱스로 설정하여 엑셀에서 그룹으로 묶이게 만듭니다.
final_excel_format = final_melted.set_index(['구분', '년도', '범죄종류'])


# 11. 엑셀에서 열었을 때 가장 깨끗하게 보이는 인코딩(utf-8-sig)으로 저장
final_excel_format.to_csv("경찰청_사이버범죄_대그룹_원하는모양.csv", encoding='utf-8-sig')

# 결과 출력
display(final_excel_format)
```

```python
import pandas as pd

# 1. 파일 이름 정의
file1 = '경찰청_연도별 사이버 범죄 통계 현황_20231231.csv'
file2 = '경찰청_사이버사기범죄 현황_20241231.csv'

# 2. 데이터 로드 (한글 깨짐 방지를 위해 EUC-KR 적용)
df1 = pd.read_csv(file1, encoding='EUC-KR')
df2 = pd.read_csv(file2, encoding='EUC-KR')

# 3. 컬럼명 앞뒤의 공백 제거
df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

# 4. 2017~2023년 데이터 필터링
df1_filtered = df1[(df1['연도'] >= 2017) & (df1['연도'] <= 2023)].copy()
df2_filtered = df2[(df2['연도'] >= 2017) & (df2['연도'] <= 2023)].copy()

# 5. 두 파일의 데이터를 하나로 결합
df_filtered = pd.concat([df1_filtered, df2_filtered], ignore_index=True)

# 6. '구분' 컬럼의 데이터 자체 앞뒤 공백 청소 및 불필요한 행(연애/투자) 제외
df_filtered['구분'] = df_filtered['구분'].astype(str).str.strip()
df_filtered = df_filtered[~df_filtered['구분'].str.contains('연애|연예|투자', na=False)].copy()

# 7. 두 파일의 데이터가 겹쳐서 수치가 부풀려지거나 중복 행이 생기지 않도록 원천 합산(통합)
df_grouped = df_filtered.groupby(['연도', '구분']).sum().reset_index()


# 8. 대그룹으로 묶을 컬럼 리스트 정의
hacking_cols = ['해킹_계정도용', '해킹_단순침입', '해킹_자료유출', '해킹_자료훼손']
scam_cols = ['직거래', '쇼핑몰', '게임', '이메일 무역', '사이버사기_직거래', '사이버사기_쇼핑몰', '사이버사기_게임', '사이버사기_이메일 무역사기']
cyber_cols = ['사이버금용범죄_피싱', '사이버금용범죄_메신저이용사기', '사이버금용범죄_기타', '사이버금융범죄_피싱', '사이버금융범죄_메신저이용사기', '사이버금융범죄_기타',
              '사이버 음란물_아동음란물', '사이버 음란물_불법촬영물유포', '사이버 도박_스포츠토토', '사이버 도박_경마 경륜 경정',
              '사이버 도박_카지노', '사이버 도박_기타', '사이버 명예훼손_모욕', '사이버 스토킹']
Others_cols = ['사이버사기_기타', '기타 불법 컨텐츠']

# 데이터프레임에 실제로 존재하는 컬럼만 필터링
hacking_present = [c for c in hacking_cols if c in df_grouped.columns]
scam_present = [c for c in scam_cols if c in df_grouped.columns]
cyber_present = [c for c in cyber_cols if c in df_grouped.columns]
Others_present = [c for c in Others_cols if c in df_grouped.columns]


# 9. 대그룹별 합산 데이터를 담을 가로형 데이터프레임 생성
mixing_df = pd.DataFrame()
mixing_df['년도'] = df_grouped['연도']
mixing_df['구분'] = df_grouped['구분']

mixing_df['Hacking'] = df_grouped[hacking_present].fillna(0).sum(axis=1) if hacking_present else 0
mixing_df['Scam'] = df_grouped[scam_present].fillna(0).sum(axis=1) if scam_present else 0
mixing_df['Cyber'] = df_grouped[cyber_present].fillna(0).sum(axis=1) if cyber_present else 0
mixing_df['Others'] = df_grouped[Others_present].fillna(0).sum(axis=1) if Others_present else 0


# 10. ⭐ 원하시는 엑셀 이미지 구조로 세로 변환 (Melt 처리) ⭐
final_melted = mixing_df.melt(
    id_vars=['구분', '년도'],
    value_vars=['Hacking', 'Scam', 'Cyber', 'Others'],
    var_name='범죄종류',
    value_name='값'
)

# 11. 컬럼 배치 순서 조정 (구분 | 년도 | 범죄종류 | 값)
final_melted = final_melted[['구분', '년도', '범죄종류', '값']]

# 12. 최종 정렬 (년도 내림차순, 구분 오름차순, 범죄종류 오름차순)
final_melted = final_melted.sort_values(by=['년도', '구분', '범죄종류'], ascending=[False, True, True]).reset_index(drop=True)


# 13. 최종 가공 데이터 CSV 파일로 저장
final_melted.to_csv("경찰청_사이버범죄_대그룹_최종통합_세로형.csv", index=False, encoding='EUC-KR')

# 최종 결과물 출력
display(final_melted)
```

```python
import pandas as pd

def safe_read_csv(file_path):
    """인코딩 에러를 방지하며 안전하게 CSV를 읽어오는 함수"""
    encodings = ['utf-8', 'cp949', 'utf-8-sig']
    for enc in encodings:
        try:
            return pd.read_csv(file_path, encoding=enc)
        except UnicodeDecodeError:
            continue
    raise ValueError(f"{file_path} 파일을 읽을 수 없습니다. 인코딩을 확인하세요.")

def run_simple_pipeline(monthly_file, content_file):
    print("🔄 데이터 파이프라인 가동 시작...")

    # ==========================================
    # 1. 월별 데이터 전처리
    # ==========================================
    try:
        df_monthly_raw = safe_read_csv(monthly_file)

        # 1월~12월 컬럼을 가로에서 세로(Row) 형태로 녹여내기
        month_cols = [f'{i}월' for i in range(1, 13)]
        melted_monthly = df_monthly_raw.melt(
            id_vars=['연도', '구분'],
            value_vars=month_cols,
            var_name='월',
            value_name='건수'
        )

        # '발생건수'와 '검거건수'를 별도의 컬럼으로 분리
        df_monthly_clean = melted_monthly.pivot_table(
            index=['연도', '월'],
            columns='구분',
            values='건수',
            aggfunc='sum'
        ).reset_index()

        # 정렬용 숫자 및 검거율 계산
        df_monthly_clean['월_숫자'] = df_monthly_clean['월'].str.replace('월', '').astype(int)
        df_monthly_clean['검거율(%)'] = (df_monthly_clean['검거건수'] / df_monthly_clean['발생건수'] * 100).round(2)
        df_monthly_clean = df_monthly_clean.sort_values(['연도', '월_숫자']).reset_index(drop=True)

        # 결과 저장
        df_monthly_clean.to_csv('정제된_월별_데이터.csv', index=False, encoding='utf-8-sig')
        print("✅ 1. 월별 데이터 정제 완료 -> '정제된_월별_데이터.csv' 저장됨")

    except Exception as e:
        print(f"❌ 월별 데이터 처리 중 오류 발생: {e}")

    # ==========================================
    # 2. 내용별(대그룹) 데이터 전처리
    # ==========================================
    try:
        df_content_raw = safe_read_csv(content_file)

        # 영문 컬럼명 한글로 표준화
        df_content_raw = df_content_raw.rename(columns={'Year': '연도', 'Type': '구분'})

        # 범죄 유형 컬럼들 추출 (Hacking, Scam, Cyber, Others 등)
        crime_types = [col for col in df_content_raw.columns if col not in ['연도', '구분']]

        # 가로 데이터를 세로 형태로 변환
        melted_content = df_content_raw.melt(
            id_vars=['연도', '구분'],
            value_vars=crime_types,
            var_name='범죄유형',
            value_name='건수'
        )

        # '발생건수'와 '검거건수'를 별도의 컬럼으로 분리
        df_content_clean = melted_content.pivot_table(
            index=['연도', '범죄유형'],
            columns='구분',
            values='건수',
            aggfunc='sum'
        ).reset_index()

        # 검거율 계산
        df_content_clean['검거율(%)'] = (df_content_clean['검거건수'] / df_content_clean['발생건수'] * 100).round(2)
        df_content_clean = df_content_clean.sort_values(['연도', '범죄유형']).reset_index(drop=True)

        # 결과 저장
        df_content_clean.to_csv('정제된_범죄내용별_데이터.csv', index=False, encoding='utf-8-sig')
        print("✅ 2. 범죄내용별 데이터 정제 완료 -> '정제된_범죄내용별_데이터.csv' 저장됨")

    except Exception as e:
        print(f"❌ 범죄내용별 데이터 처리 중 오류 발생: {e}")

# ==========================================
# 실행부 (파일명만 본인 파일명과 똑같이 맞춰주세요)
# ==========================================
monthly_csv_path = '경찰청_월별 사이버범죄 발생건수와 검거건수 현황_20231231 (1).csv'
content_csv_path = '경찰청_사이버범죄_대그룹_통합_추출.csv'

run_simple_pipeline(monthly_csv_path, content_csv_path)
```

```python
import pandas as pd

def read_csv(file_path):
    """인코딩 문제를 방지하며 안전하게 CSV를 읽어오는 함수"""
    encodings = ['utf-8', 'cp949', 'utf-8-sig']
    for enc in encodings:
        try:
            return pd.read_csv(file_path, encoding=enc)
        except UnicodeDecodeError:
            continue
    raise ValueError(f"{file_path} 파일을 읽을 수 없습니다. 인코딩을 확인하세요.")

def clean_value(val):
    """숫자 데이터에 포함된 쉼표(,)를 제거하고 숫자로 변환하는 함수"""
    if pd.isna(val):
        return 0
    if isinstance(val, str):
        val = val.replace(',', '').strip()
    try:
        return int(float(val))
    except ValueError:
        return 0

def run_2017_pipeline(monthly_file, content_file):
    print("🔄 2017년 기준 데이터 파이프라인 가동 시작...")

    # ==========================================
    # 1. 월별 데이터 전처리 (2017년 이후 필터링)
    # ==========================================
    try:
        df_monthly_raw = safe_read_csv(monthly_file)

        # '연도' 컬럼을 숫자형으로 안전하게 변환
        df_monthly_raw['연도'] = df_monthly_raw['연도'].astype(int)

        # 🌟 2017년 이상인 데이터만 필터링
        df_monthly_raw = df_monthly_raw[df_monthly_raw['연도'] >= 2017]

        # 1월~12월 컬럼 추출 및 가로 데이터를 세로로 변환 (Melting)
        month_cols = [f'{i}월' for i in range(1, 13)]

        # 문자열 쉼표 제거 및 숫자 변환 적용
        for col in month_cols:
            df_monthly_raw[col] = df_monthly_raw[col].apply(clean_value)

        melted_monthly = df_monthly_raw.melt(
            id_vars=['연도', '구분'],
            value_vars=month_cols,
            var_name='월',
            value_name='건수'
        )

        # 발생건수와 검거건수를 컬럼으로 피벗
        df_monthly_clean = melted_monthly.pivot_table(
            index=['연도', '월'],
            columns='구분',
            values='건수',
            aggfunc='sum'
        ).reset_index()

        # 월 정렬 및 검거율 계산
        df_monthly_clean['월_숫자'] = df_monthly_clean['월'].str.replace('월', '').astype(int)
        df_monthly_clean['검거율(%)'] = (df_monthly_clean['검거건수'] / df_monthly_clean['발생건수'] * 100).round(2)
        df_monthly_clean['검거율(%)'] = df_monthly_clean['검거율(%)'].fillna(0) # 0 나누기 방지

        df_monthly_clean = df_monthly_clean.sort_values(['연도', '월_숫자']).reset_index(drop=True)

        # 컬럼 순서 이쁘게 정렬 및 저장
        df_monthly_clean = df_monthly_clean[['연도', '월', '월_숫자', '발생건수', '검거건수', '검거율(%)']]
        df_monthly_clean.to_csv('정제된_월별_데이터_2017이후.csv', index=False, encoding='utf-8-sig')
        print("✅ 1. 월별 데이터 정제 완료 -> '정제된_월별_데이터_2017이후.csv' 저장됨")

    except Exception as e:
        print(f"❌ 월별 데이터 처리 중 오류 발생: {e}")

    # ==========================================
    # 2. 내용별(대그룹) 데이터 전처리 (2017년 이후 필터링)
    # ==========================================
    try:
        df_content_raw = safe_read_csv(content_file)

        # 컬럼명 표준화 (Year -> 연도, Type -> 구분)
        df_content_raw = df_content_raw.rename(columns={'Year': '연도', 'Type': '구분'})
        df_content_raw['연도'] = df_content_raw['연도'].astype(int)

        # 🌟 2017년 이상인 데이터만 필터링
        df_content_raw = df_content_raw[df_content_raw['연도'] >= 2017]

        # 범죄 유형 컬럼들 (Hacking, Scam, Cyber, Others 등) 숫자 변환 적용
        crime_types = [col for col in df_content_raw.columns if col not in ['연도', '구분']]
        for col in crime_types:
            df_content_raw[col] = df_content_raw[col].apply(clean_value)

        # 가로 데이터를 세로 형태로 변환
        melted_content = df_content_raw.melt(
            id_vars=['연도', '구분'],
            value_vars=crime_types,
            var_name='범죄유형',
            value_name='건수'
        )

        # 발생건수와 검거건수를 컬럼으로 피벗
        df_content_clean = melted_content.pivot_table(
            index=['연도', '범죄유형'],
            columns='구분',
            values='건수',
            aggfunc='sum'
        ).reset_index()

        # 검거율 계산
        df_content_clean['검거율(%)'] = (df_content_clean['검거건수'] / df_content_clean['발생건수'] * 100).round(2)
        df_content_clean['검거율(%)'] = df_content_clean['검거율(%)'].fillna(0)

        df_content_clean = df_content_clean.sort_values(['연도', '범죄유형']).reset_index(drop=True)
        df_content_clean = df_content_clean[['연도', '범죄유형', '발생건수', '검거건수', '검거율(%)']]

        # 결과 저장
        df_content_clean.to_csv('정제된_범죄내용별_데이터_2017이후.csv', index=False, encoding='utf-8-sig')
        print("✅ 2. 범죄내용별 데이터 정제 완료 -> '정제된_범죄내용별_데이터_2017이후.csv' 저장됨")

    except Exception as e:
        print(f"❌ 범죄내용별 데이터 처리 중 오류 발생: {e}")

# ==========================================
# 실행 제어부 (파일명이 일치하는지 꼭 확인하세요!)
# ==========================================
monthly_csv_path = '경찰청_월별 사이버범죄 발생건수와 검거건수 현황_20231231 (1).csv'
content_csv_path = '경찰청_사이버범죄_대그룹_통합_추출.csv'

run_2017_pipeline(monthly_csv_path, content_csv_path)
```

```python
#머신러닝 과정과 결과
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

def safety_read_csv(file_path):
    encodings = ['utf-8', 'cp949', 'utf-8-sig']
    for enc in encodings:
        try: return pd.read_csv(file_path, encoding=enc)
        except UnicodeDecodeError: continue
    raise ValueError(f"{file_path} 파일을 읽을 수 없습니다.")

def final_value(val):
    if pd.isna(val): return 0
    if isinstance(val, str): val = val.replace(',', '').strip()
    try: return int(float(val))
    except ValueError: return 0

def run_perfect_ml_pipeline(monthly_file, content_file, predict_years=[2024, 2025, 2026]):
    print("🚀 머신러닝 예측 및 검거율 연산 파이프라인 가동...")

    # ==========================================
    # 1. 내용별(대그룹) 데이터 예측 (구조 전면 보정)
    # ==========================================
    df_content_raw = safe_read_csv(content_file)
    df_content_raw = df_content_raw.rename(columns={'Year': '연도', 'Type': '구분'})
    df_content_raw['연도'] = df_content_raw['연도'].astype(int)

    # 2017년 이후 데이터 필터링 및 쉼표 제거
    df_content_raw = df_content_raw[df_content_raw['연도'] >= 2017]
    crime_types = [col for col in df_content_raw.columns if col not in ['연도', '구분']]
    for col in crime_types:
        df_content_raw[col] = df_content_raw[col].apply(clean_value)

    # 가로 테이블 -> 세로 구조 가공
    melted_content = df_content_raw.melt(id_vars=['연도', '구분'], value_vars=crime_types, var_name='범죄유형', value_name='건수')

    # 발생/검거 피벗 및 인덱스 초기화
    df_content_clean = melted_content.pivot_table(index=['연도', '범죄유형'], columns='구분', values='건수', aggfunc='sum').reset_index()
    df_content_clean.columns.name = None

    # 실제 데이터 컬럼명 강제 지정
    df_content_clean = df_content_clean.rename(columns={'발생건수': '발생건수_실제', '검거건수': '검거건수_실제'})
    df_content_clean['검거율_실제(%)'] = np.where(df_content_clean['발생건수_실제'] > 0, (df_content_clean['검거건수_실제'] / df_content_clean['발생건수_실제'] * 100).round(2), 0.0)

    # 미래 예측 루프 수행
    future_records = []
    for c_type in df_content_clean['범죄유형'].unique():
        df_sub = df_content_clean[df_content_clean['범죄유형'] == c_type].sort_values('연도')

        # 데이터가 부족할 때를 대비해 가중치 부여 또는 기본 선형회귀 강제 매핑
        X_train = df_sub[['연도']].values
        y_inc = df_sub['발생건수_실제'].values
        y_arr = df_sub['검거건수_실제'].values

        # 모델 생성 및 학습
        model_inc = LinearRegression().fit(X_train, y_inc)
        model_arr = LinearRegression().fit(X_train, y_arr)

        for yr in predict_years:
            pred_inc_val = max(1, int(round(model_inc.predict([[yr]])[0]))) # 최소 1건 이상 나오도록 보정
            pred_arr_val = max(0, int(round(model_arr.predict([[yr]])[0])))

            # 검거건수가 발생건수보다 크게 나오는 현상 방지
            if pred_arr_val > pred_inc_val:
                pred_arr_val = int(pred_inc_val * 0.8)

            pred_rate = round((pred_arr_val / pred_inc_val * 100), 2)

            future_records.append({
                '연도': yr, '범죄유형': c_type,
                '발생건수_예측': pred_inc_val, '검거건수_예측': pred_arr_val, '예측검거율(%)': pred_rate
            })

    df_future_content = pd.DataFrame(future_records)

    # 실제 데이터와 예측 데이터 통합
    content_final = pd.concat([df_content_clean, df_future_content], ignore_index=True)
    content_final = content_final.sort_values(['연도', '범죄유형']).reset_index(drop=True)

    # 컬럼 배치 정렬
    content_cols = ['연도', '범죄유형', '발생건수_실제', '검거건수_실제', '검거율_실제(%)', '발생건수_예측', '검거건수_예측', '예측검거율(%)']
    content_final = content_final[content_cols]

    print("✅ 데이터 변환 및 머신러닝 예측 연산 완료!")
    return content_final

monthly_csv = '경찰청_월별 사이버범죄 발생건수와 검거건수 현황_20231231 (1).csv'
content_csv = '경찰청_사이버범죄_대그룹_통합_추출.csv'

# 파이프라인 실행 후 결과 변수에 담기
content_result = run_perfect_ml_pipeline(monthly_csv, content_csv)

```
